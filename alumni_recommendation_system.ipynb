```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import NearestNeighbors
import numpy as np


df = pd.read_csv("/content/drive/MyDrive/alumni_profiles.csv", encoding='ISO-8859-1')


le_gender = LabelEncoder()
le_company = LabelEncoder()
le_profession = LabelEncoder()

df['gender_enc'] = le_gender.fit_transform(df['Gender'])
df['company_enc'] = le_company.fit_transform(df['Company Name'])
df['profession_enc'] = le_profession.fit_transform(df['Working Profession'])


features = df[['Experience (Years)', 'profession_enc', 'company_enc', 'gender_enc']]

knn = NearestNeighbors(n_neighbors=11)
knn.fit(features)


sample_df = df.sample(5).reset_index(drop=True)
print("💡 Choose one alumni to follow from these:\n")
for i, row in sample_df.iterrows():
    print(f"{i}: 👤 {row['Name']} | {row['Working Profession']} | {row['Company Name']} | {row['Experience (Years)']} yrs")


selected = int(input("\nEnter the number (0–4) of the alumni you want to follow: "))
selected_name = sample_df.iloc[selected]['Name']
actual_index = df[df['Name'] == selected_name].index[0]


distances, indices = knn.kneighbors([features.iloc[actual_index]])
print(f"\n🔍 You followed: {selected_name}")
print("\n🎯 Recommended Alumni to Follow:\n")
for idx in indices[0][1:]:
    row = df.iloc[idx]
    print(f"👤 {row['Name']} | {row['Working Profession']} | {row['Company Name']} | {row['Experience (Years)']} yrs")
```
